# Preamble

```python
from pathlib import Path
import hedgehogs as hdg
```

## Iterable progress

```python
values = []
for index in hdg.Progress(range(20), desc="Evaluating"):
    values.append(index**2)
```

## Manual progress

```python
with hdg.Progress(total=10, desc="Iterations") as progress:
    for iteration in range(10):
        error = 1 / (iteration + 1)
        progress.set(error=f"{error:.4f}")
        progress.update()
```

## Status messages

```python
hdg.echo("Reading data", tone="info")
hdg.echo("Fallback selected", tone="warning")
hdg.echo("Evaluation complete", tone="success")
hdg.rule("Summary")
```

## Redirected output

```python
output_root = Path.cwd() / "output" / "cli"
output_root.mkdir(parents=True, exist_ok=True)
with (output_root / "status.log").open("w") as stream:
    hdg.echo("Evaluation complete", tone="success", file=stream)
    hdg.rule("Summary", file=stream)
print((output_root / "status.log").read_text())
```
