# Preamble

```python
from pathlib import Path

import pandas as pd
import hedgehogs as hdg
```

## Output directory

```python
table_root = Path.cwd() / "output" / "tables"
table_root.mkdir(parents=True, exist_ok=True)
```

## Data

```python
results = [("Baseline", 0.0412, 1.25), ("Candidate", 0.0184, 0.82), ("Reference", 0.0102, 0.55)]
```

## Static table

```python
table = hdg.Table(
    "Model comparison", columns=["Model", "RMSE", "Time (s)"],
    formatters={"RMSE": ".3f", "Time (s)": ".2f"},
)
for row in results:
    table.add_row(*row)
table.show()
```

## Numeric sorting

```python
table.sort("RMSE", key=float)
table.show()
```

## DataFrame

```python
summary = pd.DataFrame(results, columns=["Model", "RMSE", "Time (s)"])
dataframe_table = hdg.Table.from_dataframe(
    summary, title="Evaluation", columns=["Model", "RMSE"],
    formatters={"RMSE": ".4f"},
)
dataframe_table.show()
```

## Live table

```python
live_table = hdg.Table(
    "Iterations", columns=["Iteration", "Error"],
    mode="live", formatters={"Error": ".4f"},
)
live_table.add_row(0, 0.1)
for iteration in range(1, 6):
    live_table.update_row(0, iteration, 0.1 / (iteration + 1))
live_table.finish()
```

## Export

```python
csv_path = table.to_csv(table_root / "comparison.csv")
latex = table.to_latex(table_root / "comparison.tex", caption="Model comparison", label="tab:comparison")
print(latex)
```
